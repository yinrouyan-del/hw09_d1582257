```python
# ==========================================
# 測試一：原始資料
# ==========================================

# 1. 原始資料設定
activity_names = ("桌遊體驗", "電影欣賞")  # tuple 保存活動名稱
board_raw = " Amy, Ben, amy, Cora "     # 原始桌遊文字
movie_raw = " ben, Dora, ERIC, ben "     # 原始電影文字
unit_fees = {"桌遊體驗": 120, "電影欣賞": 80}  # dict 儲存每人單價

# 2. 拆解活動名稱 (拆包或索引)
act_board = activity_names[0]
act_movie = activity_names[1]

# 3. 字串整理：去除半形空白、轉小寫、依逗號拆分成 list (保留重複)
board_list = board_raw.replace(" ", "").lower().split(",")
movie_list = movie_raw.replace(" ", "").lower().split(",")

# 4. 集合運算：去除重複與名單比較
board_set = set(board_list)
movie_set = set(movie_list)

common_set = board_set & movie_set       # 共同參加者 (交集)
all_set = board_set | movie_set          # 全部參加者 (聯集)
only_board_set = board_set - movie_set   # 只參加桌遊 (差集)
only_movie_set = movie_set - board_set   # 只參加電影 (差集)

# 5. 排序名單與切片取得前兩位
all_sorted_list = sorted(all_set)
notice_preview = all_sorted_list[:2]     # 切片取得前兩位

# 6. 計算人數與費用，並存入 summary 字典
summary = {
    "桌遊人數": len(board_set),
    "電影人數": len(movie_set),
    "共同人數": len(common_set),
    "總不重複人數": len(all_set),
    "報名人次": len(board_set) + len(movie_set),
    "桌遊應收": len(board_set) * unit_fees[act_board],
    "電影應收": len(movie_set) * unit_fees[act_movie],
    "總應收": (len(board_set) * unit_fees[act_board]) + (len(movie_set) * unit_fees[act_movie])
}

# 7. 輸出完整摘要
print("HW09 活動報名摘要")
print(f"活動：{act_board}，每人 {unit_fees[act_board]:.2f} 元")
print(f"活動：{act_movie}，每人 {unit_fees[act_movie]:.2f} 元")
print(f"桌遊整理串列：{board_list}")
print(f"電影整理串列：{movie_list}")
print(f"桌遊去重名單：{sorted(board_set)}")
print(f"電影去重名單：{sorted(movie_set)}")
print(f"桌遊人數：{summary['桌遊人數']} 人")
print(f"電影人數：{summary['電影人數']} 人")
print(f"共同參加者：{sorted(common_set)}")
print(f"共同人數：{summary['共同人數']} 人")
print(f"只參加桌遊者：{sorted(only_board_set)}")
print(f"只參加電影者：{sorted(only_movie_set)}")
print(f"全部參加者：{all_sorted_list}")
print(f"總不重複人數：{summary['總不重複人數']} 人")
print(f"報名人次：{summary['報名人次']} 人次")
print(f"通知名單前兩位：{notice_preview}")
print(f"桌遊應收：{summary['桌遊應收']:.2f} 元")
print(f"電影應收：{summary['電影應收']:.2f} 元")
print(f"總應收：{summary['總應收']:.2f} 元")
```
